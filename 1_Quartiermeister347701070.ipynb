```python
import math
import matplotlib.pyplot as plt
import numpy as np
```

```python
a = 60
half_t = 12
```

```python
def v_t(t):
  gamma = math.log(2)/half_t
  return a*math.exp(-gamma*t)
```

```python
t_1 = 0
t_2 = 250
freq = 20 * 10^3
```

```python
t_vals = np.linspace(t_1, t_2, freq)
v_vals = [v_t(t) for t in t_vals]


half_t_voltage = v_t(half_t)

plt.figure(figsize=(10, 6))

plt.plot(t_vals, v_vals, label='$v(t) = 60 e^{-\gamma t}$')

plt.axhline(y=half_t_voltage, color='r', linestyle='--', label='half_time voltage')


plt.xlabel('t')
plt.ylabel('v(t)')
plt.title('Decay simulation')

plt.grid(True)
plt.legend()
plt.show()
```
